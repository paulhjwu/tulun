# Sort Tao verse pairs by chrF++

Reads `eval/tao/results/tao_cunp89t_vs_chang.csv` and sorts the verse pairs by `chrf_norm` (sentence-level chrF++ on normalized text) in descending order, so the most similar CUNP89T/Chang pairs come first. Writes `sorted.csv` next to this notebook.

```python
%pip install -q pandas
```

```python
from pathlib import Path

import pandas as pd

pd.set_option("display.max_colwidth", 120)

INPUT_FILE = Path("../results/tao_cunp89t_vs_chang.csv")
OUTPUT_FILE = Path("sorted.csv")
```

## 1. Load

```python
df = pd.read_csv(INPUT_FILE, dtype={"key": str})
print(f"{len(df)} pairs")
df.head(3)
```

## 2. Sort by `chrf_norm`, descending

Ties are broken by `key` so the order is stable across runs.

```python
sorted_df = (
    df.sort_values(["chrf_norm", "key"], ascending=[False, True])
      .reset_index(drop=True)
)
sorted_df[["key", "book", "chapter", "verses", "chrf_raw", "chrf_norm", "len_ratio"]].head(10)
```

```python
sorted_df[["key", "book", "chapter", "verses", "chrf_raw", "chrf_norm", "len_ratio"]].tail(10)
```

## 3. Save

```python
sorted_df.to_csv(OUTPUT_FILE, index=False)
print("saved", len(sorted_df), "rows to", OUTPUT_FILE.resolve())
```
